```python
def bethe_tail(E, A, B, C):
    E=np.array(E)
    return A*np.log(B*E)/E + C

def uncertainties_from_cov(popt,cov):
    err=[]
    print('parameters and there error')
    for n in range(len(cov)):
        err.append(np.sqrt(cov[n,n]))
        print(f'{popt[n]:.3g} ± {np.sqrt(cov[n,n]):.1g}')
    return err

def get_stopping_power_spline(mat,k=5,E_min=1e4,E_max=1e7,new_file=True):
    energies,stopping_powers = an.SR_file_write_and_read(mat,E_min=E_min,E_max=E_max,new_file=new_file)
    while k>0:    
        try:    
            spline = make_interp_spline(energies,stopping_powers,k=k,bc_type='periodic')
            break
        except:
            k-=1
            if k==1:
                spline = make_interp_spline(energies,stopping_powers,k=k)
    return spline
def get_spline_data(spline,xs,extrapolate = False,non_negative=False):
    spline_with_nan = spline(xs, extrapolate=extrapolate)
    spline_data = np.nan_to_num(spline_with_nan,nan=0.00)

    if non_negative:
        spline_data = np.clip(spline_data,0,None) #sets all negative values to zero

    return spline_data
#materials = gh.create_materials(fuel_name)
fuel_name= 'Fuel_A'
mat = gh.get_material_by_name(fuel_name,mats)
E_min=1e4
E_max=1e7
energies,stopping_powers = an.SR_file_write_and_read(mat,E_min=E_min,E_max=E_max,new_file=False)

fig,ax = plt.subplots()
ax.plot(energies,stopping_powers,'x',label=fuel_name)
ax.set_xlabel('Energy [eV]')
ax.set_ylabel('Stoppingpower [eV/Å]')    
if E_min>5e5:
    popt, cov = curve_fit(bethe_tail,energies, stopping_powers,p0=[ 3.22e+07, 2.8e-06, -1.44])
    stopping_powers_fit = bethe_tail(energies, *popt)
    ax.plot(energies,stopping_powers_fit,label='fit')
    pp.plot_residuals(energies,stopping_powers,bethe_tail,energy_lin,popt,xlabel='Energy [eV]',ylabel='Stoppingpower [eV/Å]',title='Stopping power')

    err = uncertainties_from_cov(popt,cov)

energy_lin = np.linspace(0,5e6,100)

stop_spline = get_stopping_power_spline(mat,E_min=E_min,E_max=E_max,new_file=False)
stop_spline_data = get_spline_data(stop_spline,energy_lin,extrapolate = True)

ax.plot(energy_lin,stop_spline_data,label = 'spline')

ax.legend()
print(stop_spline_data[0:10])

pp.savefig('stoppingpower_spline')
```

```python

#Also possible with .tendl files seems jendl is better but tendl is more exstensive

def get_crosssection_spline(nuc_name, mt,k=1, xs=None, x_min=2e6, x_max=8e6):
    #Getting data
    symbol,Z,A = an.isolate_atomic_symbol(nuc_name)
    ev = Evaluation(f"/root/TENDL-a/a-{symbol}{A}.tendl")
    mf=3 #Reaction cross sections

    try:
        endf_output = StringIO(ev.section[mf,mt])
    except:
        #print(f'mt={mt} does not exist')
        return None,None, None
    head = get_head_record(endf_output)
    params, tab = get_tab1_record(endf_output)
    #print('energy',tab.x)   # energies in eV
    #print('cross section',tab.y)   # cross sections in barns

    #Removing data under x_min, over x_max and duplicates 
    mask = (tab.x >= x_min) & (tab.x <= x_max)
    tab.x = tab.x[mask]
    tab.y = tab.y[mask]
    if len(tab.x) ==0:
        #print(f'no data between {x_min} eV and {x_max} eV')
        return None,None,None
    tab.x, idx = np.unique(tab.x, return_index=True)
    tab.y = tab.y[idx]

    if xs is None:
        xs = np.linspace(min(tab.x),max(tab.x),int(1e4))
    while k>0:    
        try:    
            spline = make_interp_spline(tab.x,tab.y,k=k)
            break
        except:
            k-=1
            if k==1:
                spline = make_interp_spline(tab.x,tab.y,k=k)
    return spline, tab, xs

def get_crossection_from_isotope(nuc_name,k=1,xs=None, x_min=0, x_max=8e6, plot=False,print_no_data=False):
    splines={}
    tabs={}
    failed_mts = []
    if plot:
        fig,ax = plt.subplots()
        ax.set_title(rf'{nuc_name} cross section for ($\alpha$,anything)')
        ax.set_xlabel('Energy [eV]')
        ax.set_ylabel(r'cross section [b]')
    
    for mt in np.array([4,16,17,22,23,24,25,28,29]):
        mt= int(mt)
        spline, tab ,xs = get_crosssection_spline(nuc_name,mt,k,xs,x_min,x_max)
        
        if spline != None:
            splines[mt]=spline
            tabs[mt]=tab
            if plot:
                ax.plot(tab.x,tab.y,'x-',label=f'mt={mt}')
                ax.plot(xs,spline(xs))
        else:
            failed_mts.append(mt)
    if len(failed_mts)>0 and print_no_data:  
        print(f'In {nuc_name} following mts had no data',failed_mts)
    if plot:
        ax.legend()
    return splines,tabs
    
#splines = get_crossection_from_material('Be009',x_max=9e7)
nuc_cross_splines={}
nuc_cross_tabs={}
'''
for nuc in mat.nuclides:
    splines,tabs = get_crossection_from_isotope(nuc.name,k=3,x_max=8e6,plot=False)
    nuc_cross_splines[nuc.name]=splines
    nuc_cross_tabs[nuc.name]=tabs
print('mt alpha reactions: 4=n\', 16=2n, 17=3n, 22=n+a\n23=n+3a, 24=2n+a, 25=3n+a, 28=n+p and 29=n+2a')  
nuc_name='F19'
xs=np.linspace(0,6e6,100)

fig,ax = plt.subplots()
ax.set_title(rf'{nuc_name} cross section for ($\alpha$,anything)')
ax.set_xlabel('Energy [eV]')
ax.set_ylabel(r'cross section [b]')
for mt in nuc_cross_splines[nuc_name]:
    spline = nuc_cross_splines[nuc_name][mt]
    tab = nuc_cross_tabs[nuc_name][mt]
    dat_plot,=ax.plot(tab.x,tab.y,'x')
    ax.plot(xs,spline(xs, extrapolate=False),label=f'mt={mt}',color=dat_plot.get_color())

ax.legend()
#pp.savefig('Be9_crossection')
'''
get_crosssection_spline('F19',4)
```

```python
#Making the basic integral

def neutrons_from_material(mat,E_min=0,E_max=6e6,nr_points=1000,
                           plot_result=True, plot_parts=True, plot_crosssections=True,
                          use_gaussian_spectra=False,ORNL_numbers=False):
    #Getting data
    energies = np.linspace(E_min,E_max,nr_points)
    
    mat_dens = mat.get_nuclide_atom_densities() #returns nuclide densities in atom/b-cm

    #Stopping power splines
    stopping_power_spline = get_stopping_power_spline(mat,E_min=1e4,E_max=1e7,new_file=False)
    stopping_powers=get_spline_data(stopping_power_spline,energies, extrapolate=True)
    stopping_powers = np.array(stopping_powers)*1e8 #Converting from eV/Å to eV/cm
    
    #Alpha spectrum
    alpha_decays, mass,nuclide_amount = an.alpha_decay_values_from_material(mat)
    if use_gaussian_spectra:
        alpha_spectra,energies = an.energy_spectra_gaussian(alpha_decays,x=energies)
    elif ORNL_numbers:
        mass_kg = mat.density*707921.1677922/1000 #mass for 25 ft^3 fuel
        alpha_energies  = [4.15e6, 4.19e6, 4.20e6, 4.40e6, 4.45e6, 4.47e6, 4.49e6, 4.58e6, 4.72e6, 4.77e6]
        alpha_sources = {"Fuel_A": [0.41e6, 1.40e6, 0.86e8, 0.65e8, 5.01e8, 2.13e8, 17.7e8, 1.85e10, 1.83e8, 4.74e10],
                         "Fuel_B": [0.25e6, 0.86e6, 0.53e8, 0.40e8, 3.06e8, 1.30e8, 10.8e8, 1.13e10, 1.12e8, 2.90e10],
                         "Fuel_C": [0.13e6, 0.45e8, 0.85e8, 0.63e8, 3.92e8, 2.09e8, 17.3e8, 1.45e10, 1.44e8, 3.71e10]}
        alpha_spectra = [a/mass_kg for a in alpha_sources[mat.name]]
    else:
        alpha_spectra = []
        alpha_energies = []
        for name, a_activities, a_energies,energy_devs,_  in alpha_decays:
            alpha_spectra.extend(a_activities)
            alpha_energies.extend(a_energies)
        #sorting energies
        alpha_spectra = [a/mass for _,a in sorted(zip(alpha_energies,alpha_spectra))]
        alpha_energies = sorted(alpha_energies)
    
    if plot_parts:
        isotope_color_lib = {'Li6':'lightblue','Li7':'b','F19':'r','Be9':'g'}
        mt_to_markers = {4: 'o',11: 's',16: '^',17: 'v',18: 'D',22: 'P',23: 'X',24: '*',
                         25: '<',28: '>',29: 'p',30: 'h',32: 'H',33: 'd',34: '8',35: '+',}
        mt_to_reaction = {4:  r'($\alpha$,n)', 11: r'($\alpha$,2n+d)',16: r'($\alpha$,2n)',
                        17: r'($\alpha$,3n)',18: r'($\alpha$,f)',22: r'($\alpha$,$\alpha$+n)',
                        23: r'($\alpha$,n+3$\alpha$)',24: r'($\alpha$,2n+$\alpha$)',
                        25: r'($\alpha$,3n+$\alpha$)',28: r'($\alpha$,n+p)',29: r'($\alpha$,n+2$\alpha$)',
                        30: r'($\alpha$,2n+2$\alpha$)',32: r'($\alpha$,n+d)',33: r'($\alpha$,n+t)',
                        34: r'($\alpha$,n+h)',35: r'($\alpha$,n+d+2$\alpha$)',
                       }
        fuel_name_to_linetype = {'Fuel_A':'dashed','Fuel_B':'dashdot','Fuel_C':'dotted'}
        #fig, ax = plt.subplots(1,3,figsize=(18,6))

    total_integrent = np.zeros(nr_points)
    
    for nuc in mat.nuclides:
        #Getting crosssections
        cross_splines,tab = get_crossection_from_isotope(nuc.name,k=5,x_max=8e6,plot=False)
        total_crosssection = np.zeros(nr_points)
        for mt in cross_splines:
            spline_data = get_spline_data(cross_splines[mt],energies, extrapolate=False,non_negative=True)
            
            if mt in [4,22,23,28,29]: #(a,n+anything)
                total_crosssection += spline_data
                
            elif mt in [11,16,24,30]: #(a,2n+anything)
                total_crosssection += 2*spline_data
                
            elif mt in [17,25]: #(a,3n+anything) for up to mt=30
                total_crosssection += 3*spline_data
            else:
                print(f'reaction with mt={mt} is not know')
                
            if plot_crosssections and np.sum(tab[mt].y)>1e-3 and plot_parts:
                data,=ax[0].plot(tab[mt].x,tab[mt].y,f'{mt_to_markers[mt]}',label=f'{nuc.name}:{mt_to_reaction[mt]}',color=isotope_color_lib[nuc.name])
                ax[0].plot(energies,spline_data,color=data.get_color())
                all_energies = np.linspace(0,8e6,100)
                all_energies_data =  get_spline_data(cross_splines[mt],all_energies,\
                                                extrapolate=False,non_negative=True)
                ax[0].plot(all_energies,all_energies_data,'--',color=data.get_color())

        
        #Making integrent
        target_density = mat_dens[nuc.name] #atom/b-cm 
        
        #total_crosssection is in barns
        #stopping powers is in eV/cm
        
        integrent = target_density*total_crosssection/stopping_powers
        if np.sum(total_crosssection)>0 and plot_parts:
            ax[1].plot(energies,integrent,label=f'{mat.name}:{nuc.name}',color=isotope_color_lib[nuc.name],linestyle=f'{fuel_name_to_linetype[mat.name]}')
            #print(f'target_density for {nuc.name} in {mat.name} = {target_density:.3g} and integrent sum {sum(integrent):.3g}')

        #print('integrent',sum(integrent),nuc.name)
        
        total_integrent += integrent

    #dx=(E_max-E_min)/nr_points
    neutrons_total = 0
    neutrons_added = []
    integrent_spline = make_interp_spline(energies,total_integrent,k=5)
    
    for E_a,alpha in zip(alpha_energies,alpha_spectra):
        result,err = integrate.quad(integrent_spline,0,E_a)
        neutrons = alpha*result
        neutrons_total +=neutrons
        neutrons_added.append(neutrons)
        
    if plot_parts:
        #ax[0].plot(energies,stopping_powers,label=f'{mat.name}:stopping power',linestyle=f'{fuel_name_to_linetype[mat.name]}')
        #ax[1].plot(energies,total_crosssection,label='cross_section')
        ax[1].plot(energies,integrent_spline(energies),label=f'{mat.name}:total')
        #ax[2].semilogy()
        #ax[2].set_xlim(3.7e6,5e6)
        #ax[0].set_ylabel('eV/cm')
        ax[0].set_ylabel('b')
        ax[1].set_ylabel('integrant')
        for a in ax:
            a.set_xlabel('Energy [eV]')
            a.legend()
            #ax1 = a.twinx()
            if use_gaussian_spectra:
                ax1.plot(energies,alpha_spectra,label='spectra',color='k',alpha=0.3)
            #else:
                #line, =ax1.plot(alpha_energies, alpha_spectra, "_", label=name,ms=10)
                #ax1.vlines(alpha_energies,[0]*len(alpha_energies),alpha_spectra,line.get_color(),alpha=0.3)
            #ax1.semilogy()
            #ax1.set_ylim(1e-2,1e7)
            #ax1.set_ylabel(r'$\alpha$/s')
        
        pp.savefig(f'part_for_integral_{mat.name}')

        #Ploting result
        fig2,ax2 = plt.subplots()
        ax3=ax2.twinx()
    
        
        if use_gaussian_spectra:
            ax3.plot(energies,alpha_spectra,label='spectra',color='k',lw=2,alpha=0.3)
            ax2.plot(energies,neutrons_added,'x',label='n')
        else:
            line, = ax3.plot(alpha_energies, alpha_spectra, "_", label=r'$\alpha$',ms=10)
            ax3.vlines(alpha_energies,[0]*len(alpha_energies),alpha_spectra,line.get_color(),alpha=0.3)
            line2, = ax2.plot(alpha_energies,neutrons_added,'rx',label='n')
        
        ax2.set_title(f'Neutrons and alpha spectra from {mat.name}')
        ax2.set_xlabel('Energy [eV]')
        ax2.set_ylabel(r'$\frac{n}{kg\cdot s}$')
        ax3.set_ylabel(r'$\frac{\alpha}{kg\cdot s}$')
        ax2.legend([line, line2],[line.get_label(), line2.get_label()],loc='upper left')
    
        
        ax2.semilogy()
        ax3.semilogy()
        
        #ax2.set_xlim(3.5e6,5e6)
        ax2.set_ylim(1e-8,1e13)
        ax3.set_ylim(1e-8,1e13)
        
        #pp.savefig(f'integral_{mat.name}')
    if use_gaussian_spectra:
        neutrons_total_integral = integrate.simpson(neutrons_added,alpha_energies)
        print('Number of alpha',integrate.simpson(alpha_spectra,energies))
        print(f'Number of neutrons pr second from {mat.name} using integral is {neutrons_total_integral:.5g}')
    total_alphas = sum(alpha_spectra)
    print(f' Alpha per second: {total_alphas:.3g}')
    neutrons_per_alpha = neutrons_total*(1e6/total_alphas)
    print(f'n/(kg*s) from {mat.name} is {neutrons_total:.7g} with {nr_points:.2g} points giving {neutrons_per_alpha:.3g} n/10^6 a')
    print(f'for {alpha_energies[-1]/1e6:.3g} MeV the neutrons pr alpha is {neutrons_added[-1]/alpha_spectra[-1]*1e6:.3g}')
    return neutrons_total, total_alphas

    
fig, ax = plt.subplots(1,2,figsize=(10,5))
points_power = 3
fuel_name= 'Fuel_A'
mat = gh.get_material_by_name(fuel_name,mats)
neutrons,_ = neutrons_from_material(mat,E_max=5e6,nr_points=10**points_power,plot_parts=True)
mass_kg = mat.density*707921.1677922/1000
print(f'{mat.name} mass in kg: {mass_kg:.3g} which gives {neutrons*mass_kg:.3g} neutrons/s\n')

fuel_name= 'Fuel_B'
mat = gh.get_material_by_name(fuel_name,mats)
neutrons,_ = neutrons_from_material(mat,E_max=5e6,nr_points=10**points_power,plot_parts=False,plot_crosssections=False)
mass_kg = mat.density*707921.1677922/1000
print(f'{mat.name} mass in kg: {mass_kg:.3g} which gives {neutrons*mass_kg:.3g} neutrons/s\n')


fuel_name= 'Fuel_C'
mat = gh.get_material_by_name(fuel_name,mats)
neutrons,_ = neutrons_from_material(mat,E_max=5e6,nr_points=10**points_power,plot_parts=False,plot_crosssections=False)
mass_kg = mat.density*707921.1677922/1000
print(f'{mat.name} mass in kg: {mass_kg:.3g} which gives {neutrons*mass_kg:.3g} neutrons/s\n')
fuel_name= 'Fuel_A'
#pp.savefig('all_fuel_comp',figure=fig)
```

```python
isotope_color_lib = {'Li6':'lightblue','Li7':'b','F19':'r','Be9':'g'}
def neutron_energy_alpha_n(E_a,theta,nuclide_name):
    m_a = 4.001506179127#openmc.data.atomic_mass('He4') #alpha mass
    m_n = 1.008664915904 #neutron mass
    m_T = openmc.data.atomic_mass(nuclide_name)#target nuclide
    
    Z, A, m =openmc.data.zam(nuclide_name)
    new_Z = Z+2
    m_R = openmc.data.atomic_mass(f'{openmc.data.ATOMIC_SYMBOL[new_Z]}{A+3}') #residual recoil nucleus
    
    Q_i =m_a+m_T-m_n-m_R  #From energy/mass conservation in m_T+m_a = m_n-m_R
    
    term0 = m_a*m_n*E_a*np.pow(np.cos(theta),2)/np.pow(m_n+m_R,2)
    term1 = (m_R*Q_i+(m_R-m_a)*E_a)/(m_a+m_R)
    term2 = 2*np.cos(theta)/(m_n+m_R)*np.sqrt(m_a*m_n*E_a*(m_R*Q_i+(m_R-m_a)*E_a)/(m_n+m_R))
    
    return term0 + term1 + term2

def transfer_kernel(E_a,E_n,nuclide_name):
    E_n_max = neutron_energy_alpha_n(E_a,0,nuclide_name)
    E_n_min = min([neutron_energy_alpha_n(E_a,np.pi/2,nuclide_name),neutron_energy_alpha_n(E_a,np.pi,nuclide_name)])

    if E_n < E_n_max and E_n > E_n_min:
        return 1/(E_n_max-E_n_min)
    else:
        return 0
fig, ax = plt.subplots()
#ax.semilogy()
#ax.set_ylim(4e-7,5.5e-7)
#ax.set_xlim(-0.1e7,0.2e7)

lin = np.linspace(0,5e6,500)
E_as = np.sort([4108600.0, 4150600.0, 4276100.0, 4603500.0, 4722400.0, 4774600.0, 3975854.0, 4077101.0, 4150000.0,
        4214700.0, 4219000.0, 4219928.0, 4271000.0, 4279693.0, 4287360.0, 4295000.0, 4325795.0, 4328547.0, 
        4362362.0, 4366100.0, 4381431.0, 4397800.0, 4414000.0, 4435000.0, 4502000.0, 4556000.0, 4596400.0, 
        4168000.0, 4332000.0, 4445000.0, 4494000.0, 4038000.0, 4151000.0, 4198000.0])
theta = np.linspace(0,2*np.pi,100)
#print(E_as)
for nuc_name in ['Li7','Be9','F19']:
    for E_a in [E_as[0],E_as[-1]]:
        print(E_a)
        #ax.plot(theta,neutron_energy_alpha_n(E_a,theta,'Be9'),label=f'$E_a$={E_a:.3g}')
        kernel = [transfer_kernel(E_a,n,nuc_name) for n in lin]
        if E_a ==E_as[-1]:
            ax.plot(lin,kernel,color = isotope_color_lib[nuc_name],label = nuc_name)
        else:
            ax.plot(lin,kernel,color = isotope_color_lib[nuc_name])
        
ax.legend()
ax.set_xlabel('Energy [eV]')
ax.set_ylabel('Probability []')
pp.savefig('kernel_plot',figure = fig)
```

```python
from concurrent.futures import ProcessPoolExecutor
from itertools import repeat


def integrent(E_a,E_n,integrent_splines,mat):
    result = 0
    for spline,nuc in zip(integrent_splines,mat.nuclides):        
        result += spline(E_a)*transfer_kernel(E_a,E_n,nuclide_name=nuc.name)
    return result

def compute_spectrum1(E_n,alpha_energies, alpha_spectra, integrent_splines, mat):#,alpha_energies,alpha_spectra,integrent_splines,mat):
    #print(f'energy: {E_n:.3g}')
    neutrons_total = 0
    for E_a,alpha in zip(alpha_energies,alpha_spectra):
        #integrent_spline = make_interp_spline(energies,total_integrent,k=5)
        result,err = integrate.quad(integrent,0,E_a,args = (E_n,integrent_splines,mat))
        neutrons = alpha*result
        neutrons_total +=neutrons
    return neutrons_total

def neutron_spectra_from_material1(mat,E_min=0,E_max=6e6,nr_points=1000,nr_energies=100,
                           plot_result=True, plot_parts=True, plot_crosssections=True,
                          use_gaussian_spectra=False,ORNL_numbers=False):
    #Getting data
    energies = np.linspace(E_min,E_max,nr_points)
    
    mat_dens = mat.get_nuclide_atom_densities() #returns nuclide densities in atom/b-cm

    #Stopping power splines
    stopping_power_spline = get_stopping_power_spline(mat,E_min=0,E_max=E_max,new_file=False)
    stopping_powers=get_spline_data(stopping_power_spline,energies, extrapolate=True)
    stopping_powers = np.array(stopping_powers)*1e8 #Converting from eV/Å to eV/cm
    
    #Alpha spectrum
    alpha_decays, mass,nuclide_amount = an.alpha_decay_values_from_material(mat)
    if use_gaussian_spectra:
        alpha_spectra,energies = an.energy_spectra_gaussian(alpha_decays,x=energies)
    else:
        alpha_spectra = []
        alpha_energies = []
        for name, a_activities, a_energies,energy_devs,_  in alpha_decays:
            alpha_spectra.extend(a_activities)
            alpha_energies.extend(a_energies)
        #sorting energies
        alpha_spectra = [a/mass for _,a in sorted(zip(alpha_energies,alpha_spectra))]
        alpha_energies = sorted(alpha_energies)
    
    if plot_parts:
        fig,ax = plt.subplots()
        ax2 = ax.twinx()
        isotope_color_lib = {'Li6':'lightblue','Li7':'b','F19':'r','Be9':'g','Zr91':'y','Zr94':'yellow','Zr90':'gold','Zr92':'goldenrod','Zr96':'orange'}
        fuel_name_to_linetype = {'Fuel_A':'dashed','Fuel_B':'dashdot','Fuel_C':'dotted'}

    integrent_splines = []
    
    for nuc in mat.nuclides:
        #Getting crosssections
        cross_splines,tab = get_crossection_from_isotope(nuc.name,k=5,x_max=E_max,plot=False)
        total_crosssection = np.zeros(nr_points)
        for mt in cross_splines:
            spline_data = get_spline_data(cross_splines[mt],energies, extrapolate=False,non_negative=True)
            
            if mt in [4,22,23,28,29]: #(a,n+anything)
                total_crosssection += spline_data
                
            elif mt in [11,16,24,30]: #(a,2n+anything)
                total_crosssection += 2*spline_data
                
            elif mt in [17,25]: #(a,3n+anything) for up to mt=30
                total_crosssection += 3*spline_data
            else:
                print(f'reaction with mt={mt} is not know')

        #Making integrent
        target_density = mat_dens[nuc.name] #atom/b-cm 
        
        #total_crosssection is in barns
        #stopping powers is in eV/cm
        
        integrent_splines.append(make_interp_spline(energies,target_density*total_crosssection/stopping_powers,k=5))
        
        if np.sum(total_crosssection)>1e-1 and plot_parts:
            ax2.plot(energies,target_density*total_crosssection/stopping_powers,label=f'{mat.name}:{nuc.name}',color=isotope_color_lib[nuc.name],linestyle=f'{fuel_name_to_linetype[mat.name]}')
            ax2.set_ylabel('integrent without kernel')
            ax2.legend(loc='upper right')
            #print(f'target_density for {nuc.name} in {mat.name} = {target_density:.3g} and integrent sum {sum(integrent):.3g}')

    neutrons_total = 0
    neutron_spectra = []
    neutron_energies =  np.linspace(E_min,E_max,nr_energies)
    #print('setup done')



    '''
    for E_n in neutron_energies:
        neutrons_total = 0
        print(f'energy: {E_n:.3g}')
        for E_a,alpha in zip(alpha_energies,alpha_spectra):
            #integrent_spline = make_interp_spline(energies,total_integrent,k=5)
            result,err = integrate.quad(integrent,0,E_a,args = (E_n,integrent_splines,mat))
            neutrons = alpha*result
            neutrons_total +=neutrons
        neutron_spectra.append(neutrons_total)'''


    with ProcessPoolExecutor() as executor:
        neutron_spectra = list(executor.map(compute_spectrum1,
            neutron_energies,
            repeat(alpha_energies),
            repeat(alpha_spectra),
            repeat(integrent_splines),
            repeat(mat),))
        
    if plot_parts:
        for E_a in [E_as[0],E_as[-1]]:
            integrent_plot = [integrent(E_a,E_n,integrent_splines,mat) for E_n in energies]
            ax.plot(energies,integrent_plot,label=f'{mat.name}:total for {E_a:.3g}')
        ax.set_ylabel('integrent total')
        ax.set_xlabel('Energy [eV]')
        ax.legend(loc='upper left')
        #pp.savefig(f'part_for_integral_{mat.name}')

    return neutron_spectra,neutron_energies
#resolution = 500

resolutions_bench = [100*np.pow(2,n) for n in range(1)]
print(resolutions_bench)
timings_bench = []

for resolution in resolutions_bench:
    start = time.perf_counter()
    
    n_spectra_bench, n_energies_bench = neutron_spectra_from_material1(mat,0,5e6,nr_points=10000,nr_energies=int(resolution),plot_parts=False)
    end = time.perf_counter()

    time_meas = end-start
    print(f"Execution time: {time_meas:.6f} seconds, resolution {resolution}")
    n_count = np.trapezoid(n_spectra_bench,n_energies_bench)
    print(f'number of neutrons {n_count:.6g}')
    timings_bench.append(time_meas)
    fig3, ax3 =plt.subplots()
      
    ax3.set_title(f'Neutron spectra from {mat.name}\n resolution: {resolution/5e3} points pr keV')
    ax3.set_xlabel('Energy [eV]')
    ax3.set_ylabel(r'$\frac{n}{kg\cdot s}$')
    ax3.plot(n_energies_bench,n_spectra_bench,'-',label = 'neutron spectra')
    
#ax3.set_xlim(2e6,3e6)
#ax3.set_ylim(8e-5,14e-5)

#pp.savefig(f'Neutron spectra from {mat.name} resolution {resolution/5e3} time {time_meas:.3g} s',figure=fig3)
```

```python
class Nuclide_integrent:
    def __init__(self, E_as, nuclide_name,spline):
        self.nuclide_name = nuclide_name

        self.E_n_maxs = [neutron_energy_alpha_n(E_a, 0, nuclide_name) for E_a in E_as]
        self.E_n_mins = [min(
            neutron_energy_alpha_n(E_a, np.pi / 2, nuclide_name),
            neutron_energy_alpha_n(E_a, np.pi, nuclide_name),
        ) for E_a in E_as]
        self.values = [spline(E_a) / (E_n_max - E_n_min) for E_a,E_n_max,E_n_min in 
                      zip(E_as,self.E_n_maxs,self.E_n_mins)]
    def __call__(self, E_n,i):
        if self.E_n_mins[i] < E_n < self.E_n_maxs[i]:
            return self.values[i]
        else:
            return 0  

def init_worker(indexs, alpha_S, E_as, integrents):
    global G_indexs, G_alpha_S, G_E_as, G_integrents
    G_indexs = indexs
    G_alpha_S = alpha_S
    G_E_as = E_as
    G_integrents = integrents


def compute_spectrum(E_n):#,indexs,alpha_spectra,energies_with_alpha,integrents):
    neutrons_total = 0
    values = [np.sum([integrent(E_n,i) for integrent in G_integrents]) for i in range(len(G_E_as))]
    
    for i,alpha in zip(G_indexs,G_alpha_S): #over all different alpha energies

        neutrons_total +=  alpha*np.trapezoid(values[:i], G_E_as[:i]) #neutrons
    return neutrons_total
        
def neutron_spectra_from_material(mat,E_min=1e4,E_max=6e6,nr_points=10000,nr_energies=100,printing=True,new_file = False):

    #Getting data
    time_setup_start = time.perf_counter()
    energies = np.linspace(E_min,E_max,nr_points)
    
    mat_dens = mat.get_nuclide_atom_densities() #returns nuclide densities in atom/b-cm

    #Stopping power splines
    stopping_power_spline = get_stopping_power_spline(mat,E_min=0,E_max=E_max,new_file=new_file)
    stopping_powers=get_spline_data(stopping_power_spline,energies, extrapolate=True)
    stopping_powers = np.array(stopping_powers)*1e8 #Converting from eV/Å to eV/cm
    
    #Alpha spectrum
    alpha_decays, mass,nuclide_amount = an.alpha_decay_values_from_material(mat)
    alpha_spectra = []
    alpha_energies = []
    for name, a_activities, a_energies,energy_devs,_  in alpha_decays:
        alpha_spectra.extend(a_activities)
        alpha_energies.extend(a_energies)
    #sorting energies
    alpha_spectra = [a/mass for _,a in sorted(zip(alpha_energies,alpha_spectra))]
    alpha_energies = sorted(alpha_energies)
    
    integrent_splines = []
    
    for nuc in mat.nuclides:
        #Getting crosssections
        cross_splines,tab = get_crossection_from_isotope(nuc.name,k=5,x_max=E_max,plot=False)
        total_crosssection = np.zeros(nr_points)

        
        for mt in cross_splines:
            spline_data = get_spline_data(cross_splines[mt],energies, extrapolate=False,non_negative=True)
            
            if mt in [4,22,23,28,29]: #(a,n+anything)
                total_crosssection += spline_data
                
            elif mt in [11,16,24,30]: #(a,2n+anything)
                total_crosssection += 2*spline_data
                
            elif mt in [17,25]: #(a,3n+anything) for up to mt=30
                total_crosssection += 3*spline_data
            else:
                print(f'reaction with mt={mt} is not know')
            
        #Making integrent
        target_density = mat_dens[nuc.name] #atom/b-cm 
        
        #total_crosssection is in barns
        #stopping powers is in eV/cm
        
        integrent_splines.append(make_interp_spline(energies,target_density*total_crosssection/stopping_powers,k=5))

    #Setting energy array with all alpha energies and energies in between up
    energies_with_alpha = np.unique(np.concatenate((energies,alpha_energies)))
    indexs = [np.searchsorted(energies_with_alpha, E_a, side="right") for E_a in alpha_energies]

    #Making kernels and integrent 
    numeric_integrents = [Nuclide_integrent(energies_with_alpha,nuc.name,spline) 
                          for nuc,spline in zip(mat.nuclides,integrent_splines)]
            

    time_setup = time.perf_counter() - time_setup_start
    if printing:
        print(f'setup done in {time_setup} s')

    #Neutron energies
    neutrons_total = 0
    neutron_spectra = []
    neutron_energies =  np.linspace(E_min,E_max,nr_energies)

    with ProcessPoolExecutor(initializer=init_worker,
                             initargs=(indexs, alpha_spectra, energies_with_alpha, numeric_integrents)) as executor:
        neutron_spectra = list(executor.map(compute_spectrum, neutron_energies))     

    return neutron_spectra,neutron_energies

resolutions = [20*np.pow(2,n) for n in [4]] #np.linspace(100,500,5)
print(resolutions)

timings = []

for resolution in resolutions:
    start = time.perf_counter()
    
    n_spectra, n_energies = neutron_spectra_from_material(mat,1e4,5e6,nr_points=500,nr_energies=resolution)
    end = time.perf_counter()
    time_meas = end-start
    print(f"Execution time: {time_meas:.6f} seconds, resolution {resolution}")
    n_count = np.trapezoid(n_spectra,n_energies)
    print(f'number of neutrons {n_count:.6g}')
    timings.append(time_meas)

    fig3, ax3 =plt.subplots()
          
    ax3.set_title(f'Neutron spectra from {mat.name}\n resolution: {resolution/5e3} points pr keV')
    ax3.set_xlabel('Energy [eV]')
    ax3.set_ylabel(r'$\frac{n}{kg\cdot s}$')
    ax3.plot(n_energies,n_spectra,'-',label = 'neutron spectra')
```
